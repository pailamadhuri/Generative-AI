```python
#text to speech
from openai import OpenAI

speech = client.audio.speech.create(
    model="gpt-4o-mini-tts",
    voice="alloy",
    input="Our quarterly meeting covered revenue growth, hiring delays, and a product launch scheduled for next month. The team agreed to reduce scope for the first release."
)

speech.stream_to_file("speech.mp3")
print("Audio saved as speech.mp3")
```

```python
from IPython.display import Audio

Audio("speech.mp3")
```

```python
#speech to text
audio_file = open("speech.mp3", "rb")

transcript = client.audio.transcriptions.create(
    model="gpt-4o-transcribe",
    file=audio_file
)

print(transcript.text)
```
