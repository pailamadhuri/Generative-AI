```python
#summarized audio to text
import os
import getpass
api_key = getpass.getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)
from openai import OpenAI
client = OpenAI()
def transcribe_audio(file_path: str) -> str:
    with open(file_path, "rb") as audio_file:
        transcript = client.audio.transcriptions.create(
            model="gpt-4o-transcribe",
            file=audio_file
        )
    return transcript.text


def summarize_text(transcript_text: str) -> str:
    prompt = f"""
You are a helpful assistant.
Summarize the following audio transcript in clear, simple English.

Requirements:
1. First give a short paragraph summary.
2. Then give 5 bullet points with key ideas.
3. Keep it concise and easy to understand.

Transcript:
{transcript_text}
"""

    response = client.responses.create(
        model="gpt-5.4",
        input=prompt
    )

    return response.output_text.strip()


def main():
    print("Audio Transcription + Summary App")
    print("Type 'quit' to exit.\n")

    while True:
        file_path = input("Enter audio file path: ").strip()

        if file_path.lower() in {"quit", "exit"}:
            print("Goodbye.")
            break

        if not file_path:
            print("Please enter a file path.\n")
            continue

        if not os.path.exists(file_path):
            print("File not found. Please check the path.\n")
            continue

        try:
            print("\nTranscribing audio...")
            transcript_text = transcribe_audio(file_path)

            print("\n--- Full Transcript ---")
            print(transcript_text)

            print("\nGenerating summary...")
            summary = summarize_text(transcript_text)

            print("\n--- Summary ---")
            print(summary)
            print()

        except Exception as e:
            print(f"\nError: {e}\n")


if __name__ == "__main__":
    main()
```
